### Imports


```python
import sys
from pathlib import Path

import pandas as pd
import numpy as np
```

```python
# Locate project root
PROJECT_ROOT = Path.cwd()

if not (PROJECT_ROOT / "ml").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

from ml.preprocessing import preprocess_bars
```

### Loading the dataset

```python
raw_path = PROJECT_ROOT / "data" / "raw" / "META_5min.csv"

raw_df = pd.read_csv(raw_path)

print("Raw dataset shape:", raw_df.shape)
display(raw_df.head())
```

```python
print("Dataset information:")
raw_df.info()

print("\nMissing values:")
display(raw_df.isnull().sum())

print("\nData types:")
display(raw_df.dtypes)
```

### Apply preprocessing

```python
raw_df_backup = raw_df.copy(deep=True)

processed_df = preprocess_bars(raw_df)

print("Preprocessing completed!")
print("Processed dataset shape:", processed_df.shape)

display(processed_df.head())
```

```python
print("Processed data types:")
display(processed_df.dtypes)

print("\nChronological ordering:",
      processed_df["timestamp"].is_monotonic_increasing)

print("\nDuplicate timestamps:",
      processed_df["timestamp"].duplicated().sum())

print("\nRaw dataset unchanged:",
      raw_df.equals(raw_df_backup))

print("\nShape preserved:",
      raw_df.shape == processed_df.shape)
```
